```python
# 	21.Terminal Execution Task 
# Create a Python file named calculator.py containing separate functions for: 
# 		add() 
# 		subtract() 
# 		multiply() 
# 		divide() 
# Then execute the program through the terminal using: 
# python calculator.py 
# The program should accept two numbers from the user and display the results of all four operations. 
```

```python
# calculator.py
 
def add(a, b):
    return a + b
 
def subtract(a, b):
    return a - b
 
def multiply(a, b):
    return a * b
 
def divide(a, b):
    if b == 0:
        return "Error: division by zero"
    return a / b
 
def main():
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
 
    print("Addition:", add(num1, num2))
    print("Subtraction:", subtract(num1, num2))
    print("Multiplication:", multiply(num1, num2))
    print("Division:", divide(num1, num2))
 
if __name__ == "__main__":
    main()
```
