```python
# Q16.
# Create a dictionary using two lists.
# Example:
# keys = ["A","B","C"]
# values = [10,20,30]
# Expected:
# {'A':10,'B':20,'C':30}
keys = ["A", "B", "C"]
values = [10, 20, 30]
result = dict.fromkeys(keys)
for i, k in enumerate(keys):
    result[k] = values[i]
print(result)
```

```python
# Q28.
# Explain the use of
# keys()
# Explanation: Returns a view object of all the dictionary's keys. It's a live view it updates automatically if the dictionary changes 
# and is commonly used for iteration or membership checks.
d = {"a": 1, "b": 2}
print(d.keys())
```

```python
# Q29.
# Explain the use of
# values()
# Explanation: Returns a view object of all the dictionary's values, useful for aggregate operations like sum(), max(), or min() without needing the keys.
d = {"a": 1, "b": 2}
print(d.values())
```

```python
# Q30.
# Explain the use of
# items()
# Explanation: Returns a view of (key, value) tuple pairs the standard way to iterate over both keys and values together.
d = {"a": 1, "b": 2}
print(list(d.items()))
```

```python
# Q31.
# Explain the use of
# pop()
# Explanation: Removes a key and returns its value. Unlike del, pop() lets you supply a default so a missing key doesn't raise an error, and it hands you the removed value.
d = {"a": 1, "b": 2}
val = d.pop("a")
print(val, d)
print(d.pop("z", "not found"))
```

```python
# Q32.
# Explain the use of
# popitem()
# Explanation: Removes and returns the last inserted (key, value) pair as a tuple a quick way to drain a dictionary in reverse insertion order.
d = {"a": 1, "b": 2}
print(d.popitem())
print(d)
```

```python
# Q33.
# Explain the use of
# update()
# Explanation: Merges another dictionary (or key-value pairs) into the current one in place matching keys get overwritten, new keys get added.
d = {"a": 1, "b": 2}
d.update({"b": 20, "c": 3})
print(d)
```

```python
# Q34.
# Explain the use of
# setdefault()
# Explanation: Returns the value for a key if it exists; if not, inserts that key with a given default value and returns the default. 
# Handy for building dictionaries of lists without checking 'if key in d' every time.
d = {"a": 1}
print(d.setdefault("a", 99))
print(d.setdefault("b", 99))
print(d)
```

```python
# Q35.
# Explain the use of
# fromkeys()
# Explanation: A class method that builds a new dictionary from an iterable of keys, all mapped to the same given value (default None).
d = dict.fromkeys(["x", "y", "z"], 0)
print(d)
```

```python
# Q36.
# Count the frequency of each character in a string.
# Example
# Input:
# banana

# Expected Output:
# {
# 'b':1,
# 'a':3,
# 'n':2
# }
s = "banana"
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
```

```python
# Q37.
# Count the frequency of words in a sentence.
# Example
# Input
# I love Python and I love AI
# Expected Output
# {
# 'I':2,
# 'love':2,
# 'Python':1,
# 'and':1,
# 'AI':1
# }
sentence = "I love Python and I love AI"
word_count = {}
for word in sentence.split():
    word_count[word] = word_count.get(word, 0) + 1
    
print(word_count)
```

```python
# Q39.
# Find duplicate values in a dictionary
data = {"A": 10, "B": 20, "C": 10, "D": 30}
seen = {}
dupes = []
for k, v in data.items():
    if v in seen:
        dupes.append(v)
    else:
        seen[v] = k
print(dupes)
```

```python
# Q51.
# Given
# sales={
# "Jan":2500,
# "Feb":3000,
# "Mar":2700,
# "Apr":4000
# }
# Find the month with the highest sales.
sales = {"Jan": 2500, "Feb": 3000, "Mar": 2700, "Apr": 4000}
top_month = max(sales, key=sales.get)
print(f"{top_month} had the highest sales: {sales[top_month]}")
```

```python
# Q54.
# Merge dictionaries and, if duplicate keys exist, add their values.
# Example
# d1={"A":10,"B":20}
# d2={"B":15,"C":5}

# Expected Output
# {"A":10,"B":35,"C":5}
d1 = {"A": 10, "B": 20}
d2 = {"B": 15, "C": 5}
merged = dict(d1)
for k, v in d2.items():
    merged[k] = merged.get(k, 0) + v
print(merged)
```

```python
# Q55.
# Create a phonebook using dictionaries that allows the user to:
# Add a contact 
# Search for a contact 
# Update a contact 
# Delete a contact 
# Display all contacts 
# Exit the program 
phonebook = {}
 
def add_contact():
    name = input("Name: ")
    number = input("Number: ")
    phonebook[name] = number
    print(f"Added {name}.")
 
def search_contact():
    name = input("Name to search: ")
    if name in phonebook:
        print(f"{name}: {phonebook[name]}")
    else:
        print("Contact not found.")
 
def update_contact():
    name = input("Name to update: ")
    if name in phonebook:
        phonebook[name] = input("New number: ")
        print("Updated.")
    else:
        print("Contact not found.")
 
def delete_contact():
    name = input("Name to delete: ")
    if phonebook.pop(name, None) is not None:
        print("Deleted.")
    else:
        print("Contact not found.")
 
def display_contacts():
    if not phonebook:
        print("Phonebook is empty.")
    for name, number in phonebook.items():
        print(f"{name}: {number}")
 
while True:
    print("\n1.Add 2.Search 3.Update 4.Delete 5.Display 6.Exit")
    choice = input("Choose an option: ")
    if choice == "1": add_contact()
    elif choice == "2": search_contact()
    elif choice == "3": update_contact()
    elif choice == "4": delete_contact()
    elif choice == "5": display_contacts()
    elif choice == "6":
        print("Thank You!")
        break
    else:
        print("Invalid option, try again.")
```
