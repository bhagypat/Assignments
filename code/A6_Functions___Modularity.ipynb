```python
# 21.Write a Python function called calculate_average() that accepts a list of numbers and returns 	their average. Do not use NumPy. 
def calculate_average(numbers):
    total = 0
    for n in numbers:
        total += n
    return total / len(numbers)
 
print(calculate_average([2, 4, 6, 8]))
```

```python
# 22.Write a Python function called find_maximum() that accepts a list of numbers and returns the 	largest value without using Python's built-in max() function.
def find_maximum(numbers):
    largest = numbers[0]
    for n in numbers:
        if n > largest:
            largest = n
    return largest
 
print(find_maximum([3, 9, 2, 7, 5]))
```

```python
# 23.Write a modular Python program that contains separate functions to: 
# •	Read a list of numbers 
# •	Calculate the mean 
# •	Calculate the maximum 
# •
# 	Display the results 
def read_numbers():
    return [4, 8, 15, 16, 23, 42]
 
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)
 
def calculate_max(numbers):
    largest = numbers[0]
    for n in numbers:
        if n > largest:
            largest = n
    return largest
 
def display_results(numbers, mean_val, max_val):
    print("Numbers:", numbers)
    print("Mean:", mean_val)
    print("Max:", max_val)
 
nums = read_numbers()
display_results(nums, calculate_mean(nums), calculate_max(nums))
```

```python
# 24.Write a Python function called count_even() that accepts a list of integers and returns the 	number of even numbers in the list. 
def count_even(numbers):
    count = 0
    for n in numbers:
        if n % 2 == 0:
            count += 1
    return count
 
print(count_even([1, 2, 3, 4, 5, 6]))
```

```python
# 25.Write a modular Python program to calculate the accuracy of a classification model using 	two functions: 
# •	calculate_accuracy(y_true, y_pred) 
# •	display_result(accuracy) 
# Do not use Scikit-learn's built-in accuracy function. 
def calculate_accuracy(y_true, y_pred):
    correct = 0
    for t, p in zip(y_true, y_pred):
        if t == p:
            correct += 1
    return correct / len(y_true)
 
def display_result(accuracy):
    print(f"Accuracy: {accuracy*100:.2f}%")
 
y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 0]
display_result(calculate_accuracy(y_true, y_pred))
```

```python
# 26.Write a Python function that accepts a list of numbers and returns two values: mean and 	standard deviation. 
def mean_and_std(numbers):
    n = len(numbers)
    mean_val = sum(numbers) / n
    variance = sum((x - mean_val) ** 2 for x in numbers) / n
    std_val = variance ** 0.5
    return mean_val, std_val
 
print(mean_and_std([2, 4, 4, 4, 5, 5, 7, 9]))
```

```python
# 27.Write a function called normalize() that accepts a list of numerical values and performs min-
# max normalization: 


# 𝑥−𝑥𝑚𝑖𝑛


# 𝑥𝑛𝑜𝑟𝑚=
# 𝑥𝑚𝑎𝑥−𝑥𝑚𝑖𝑛

# Do not use Scikit-learn's preprocessing functions. 
def normalize(values):
    x_min = min(values)
    x_max = max(values)
    return [(x - x_min) / (x_max - x_min) for x in values]
 
print(normalize([10, 20, 30, 40, 50]))
```

```python
# 28.Write a modular Python program for a simple machine-learning preprocessing pipeline 	using separate functions: 
# 					load_data() 
# 					↓ 
# 		remove_missing_values() 
# 					↓ 
# 			normalize_data() 
# 					↓ 
# 				display_data() 
def load_data():
    return [1, 2, None, 4, 5, None, 7]
 
def remove_missing_values(data):
    return [x for x in data if x is not None]
 
def normalize_data(data):
    x_min, x_max = min(data), max(data)
    return [(x - x_min) / (x_max - x_min) for x in data]
 
def display_data(data):
    print("Processed data:", data)
 
raw = load_data()
clean = remove_missing_values(raw)
norm = normalize_data(clean)
display_data(norm)
```

```python
# 29.Write a Python function to calculate the Euclidean distance between two points without 	using any built-in distance function. 
def euclidean_distance(p1, p2):
    total = 0
    for a, b in zip(p1, p2):
        total += (a - b) ** 2
    return total ** 0.5
 
print(euclidean_distance((0, 0), (3, 4))) 
```
