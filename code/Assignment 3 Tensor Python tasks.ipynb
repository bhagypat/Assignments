```python
# Task 2 — From Vector to Matrix
# Suppose an industrial system has 4 sensors recording measurements for 5 time steps.
# Create the following data:
# Temperature:  [25, 26, 25, 27, 28]
# Pressure:     [101, 102, 101, 103, 104]
# Humidity:     [60, 62, 61, 63, 64]
# Vibration:    [0.2, 0.3, 0.2, 0.4, 0.5]
# Create a tensor with shape:
# (5, 4)
# where:
# rows    → time
# columns → sensors

# Then print:
# shape 
# number of dimensions 
# first time step 
# temperature readings 
# vibration readings 
# Think: What does each axis represent?

temperature = [25, 26, 25, 27, 28]
pressure    = [101, 102, 101, 103, 104]
humidity    = [60, 62, 61, 63, 64]
vibration   = [0.2, 0.3, 0.2, 0.4, 0.5]
 
# rows -> time (5), columns -> sensors (4)
X2 = np.array([temperature, pressure, humidity, vibration]).T  # shape (5,4)
 
print("Shape:", X2.shape)
print("Number of dimensions:", X2.ndim)
print("First time step (X2[0]):", X2[0])
print("Temperature readings (X2[:,0]):", X2[:, 0])
print("Vibration readings (X2[:,3]):", X2[:, 3])
```

```python
# Task 3 — Create a 3D Tensor
# Now suppose we monitor 10 machines.
# Each machine has:
# 5 time steps 
# 4 sensors 
# Create a tensor with shape:
# (10, 5, 4)
# Interpret the dimensions as:
# Machine × Time × Sensor
# Then answer using Python:
# What is the temperature reading of machine 3 at time step 2? 
# Extract all sensor readings of machine 5. 
# Extract all machines at time step 1. 
# Extract the readings of sensor 4 from all machines and all time steps. 

np.random.seed(0)
X3 = np.random.randint(20, 100, size=(10, 5, 4)).astype(float)  # Machine x Time x Sensor
 
print("Machine 3, time step 2, sensor(temp):", X3[2, 1, 0])
print("All sensor readings of machine 5, shape:", X3[4].shape)
print("All machines at time step 1, shape:", X3[:, 0, :].shape)
print("Sensor 4 from all machines & all time steps, shape:", X3[:, :, 3].shape)
```

```python
# Task 4 — Indexing a Tensor
# Create:
# X = np.arange(24).reshape(2, 3, 4)
# Print X.
# Now investigate:
# X[0]
# X[1]
# X[0, 0]
# X[0, 0, 0]

# Answer:
# What is the shape of X? 
# What does X[0] represent? 
# What does X[0,0] represent? 
# What does X[0,0,0] represent? 

X4 = np.arange(24).reshape(2, 3, 4)
print(X4)
print(X4[0])
print(X4[1])
print(X4[0, 0])
print(X4[0, 0, 0])
```

```python
# Task 5 — Slicing
# Using:
# X = np.arange(60).reshape(3, 4, 5)
# Perform:
# Extract the first block: X[0]
# Extract the first row from every block.
# Extract the third element from every row and every block.
# Extract the last two elements along the final dimension. Then print the resulting shapes.

X5 = np.arange(60).reshape(3, 4, 5)
 
first_block = X5[0]                    # first block
first_row_every_block = X5[:, 0, :]    # first row of every block
third_elem_every_row_block = X5[:, :, 2]   # third element of every row & block
last_two_final_dim = X5[:, :, -2:]     # last two elements along final axis
 
print("First block shape:", first_block.shape)
print("First row of every block shape:", first_row_every_block.shape)
print("Third element of every row/block shape:", third_elem_every_row_block.shape)
print("Last two elements along final dim shape:", last_two_final_dim.shape)
```

```python
# Task 7 — Batch of Images
# Now suppose you have 32 RGB images, each of size 128 × 128.
# Create a tensor with shape:
# (32, 128, 128, 3)
# using random values.
# Then:
# Extract the first image. 
# Extract the first 10 images. 
# Extract the red channel from all images. 
# Extract pixel (50, 50) from every image. 
# Print the shape after each operation. 

np.random.seed(2)
batch = np.random.randint(0, 256, size=(32, 128, 128, 3))
 
first_image     = batch[0]
first_10        = batch[:10]
red_all         = batch[:, :, :, 0]
pixel_50_50_all = batch[:, 50, 50, :]
 
print("Batch shape:", batch.shape)
print("First image shape:", first_image.shape)
print("First 10 images shape:", first_10.shape)
print("Red channel of all images shape:", red_all.shape)
print("Pixel (50,50) of every image shape:", pixel_50_50_all.shape)
```

```python
# Task 8 — Reshape
# Create:
# X = np.arange(24)
# Reshape it into:
# (4, 6)
# (2, 3, 4)
# (6, 4)
# (24,)
# For each one:
# Print the tensor. 
# Print its shape. 
# Print its number of dimensions. 
# Then try answering:
# Why does reshaping change the shape but not the total number of elements?

X8 = np.arange(24)
for shape in [(4, 6), (2, 3, 4), (6, 4), (24,)]:
    r = X8.reshape(shape)
    print(shape, "-> shape:", r.shape, "| ndim:", r.ndim)
```

```python
# Task 9 — Reshape vs Transpose
# Create:
# X = np.arange(24).reshape(2, 3, 4)
# Now perform:
# X.reshape(3, 2, 4) and X.transpose(1, 0, 2)
# Compare the results.
# Are these two operations doing the same thing?

X9 = np.arange(24).reshape(2, 3, 4)
reshaped   = X9.reshape(3, 2, 4)
transposed = X9.transpose(1, 0, 2)
 
print("Are they equal?:", np.array_equal(reshaped, transposed))
```

```python
# Task 10 — Broadcasting
# Create:
# X = np.ones((3, 4))
# and:
# b = np.array([10, 20, 30, 40])
# Calculate: X + b
# Then try: X * b
# Question: Why can a (4,) tensor be combined with a (3,4) tensor?

X10 = np.ones((3, 4))
b10 = np.array([10, 20, 30, 40])
 
add_result = X10 + b10
mul_result = X10 * b10
print(add_result)
print(mul_result)
```

```python
# Task 11 — Broadcasting with Sensor Data
# Suppose:
# X.shape = (100, 10)
# represents:
# 100 samples × 10 sensors
# The calibration offset for each sensor is:
# [1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4]
# Create the tensor and subtract the calibration offsets.
# Your output should remain:
# (100, 10)
# Goal: See broadcasting in an actual engineering problem.

np.random.seed(3)
X11 = np.random.randn(100, 10) + 5  # 100 samples x 10 sensors
offsets = np.array([1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4])
 
calibrated = X11 - offsets
print("Calibrated shape:", calibrated.shape)
```

```python
# Task 12 — Tensor Aggregation
# Create:
# X = np.random.randn(100, 20, 5)
# Interpret it as:
# 100 machines
# 20 time steps
# 5 sensors
# Calculate:
# Mean across time. 
# Mean across sensors. 
# Mean across machines. 
# Maximum sensor value across time. 
# Standard deviation across time. 
# Predict the resulting shape before running the code.

X12 = np.random.randn(100, 20, 5)  # machines, time, sensors
 
mean_time            = X12.mean(axis=1)   # average over time
mean_sensors         = X12.mean(axis=2)   # average over sensors
mean_machines        = X12.mean(axis=0)   # average over machines
max_sensor_over_time = X12.max(axis=1)
std_time             = X12.std(axis=1)
 
print(mean_time.shape, mean_sensors.shape, mean_machines.shape,
      max_sensor_over_time.shape, std_time.shape)
```

```python
# Task 13 — Normalize a Tensor
# Create sensor data:
# X = np.random.randn(100, 20, 5)
# Calculate the mean and standard deviation for each sensor.
# Normalize the tensor using:
#  xnormalized=x-μ
# Then verify:
# mean ≈ 0
# standard deviation ≈ 1
X13 = np.random.randn(100, 20, 5)
 
mu    = X13.mean(axis=(0, 1), keepdims=True)   # per-sensor mean, shape (1,1,5)
sigma = X13.std(axis=(0, 1), keepdims=True)    # per-sensor std,  shape (1,1,5)
 
X13_norm = (X13 - mu) / sigma
print("Normalized mean:", X13_norm.mean(axis=(0, 1)))
print("Normalized std: ", X13_norm.std(axis=(0, 1)))
```

```python
# Task 14 — Flattening
# Create an image:
# (28, 28, 3)
# Flatten it into:
# (2352,)
# Then create a batch of 64 images:
# (64, 28, 28, 3)
# Flatten each image while keeping the batch dimension.
# Expected shape: 	(64, 2352)
# Why shouldn't we simply flatten the entire tensor into (150528,) when working with a batch?
img14 = np.random.randint(0, 256, size=(28, 28, 3))
flat_single = img14.flatten()
print("Single image flatten shape:", flat_single.shape)
 
batch14 = np.random.randint(0, 256, size=(64, 28, 28, 3))
flat_batch = batch14.reshape(batch14.shape[0], -1)   # keep batch dim
print("Batch flatten (keep batch dim) shape:", flat_batch.shape)
```

```python
# Task 15 — Concatenation
# Suppose a model produces two feature tensors:
# Visual features:
# (32, 128)

# Sensor features:
# (32, 64)
# Combine them into a single feature tensor.
# Expected result:
# (32, 192)
# Use tensor concatenation.
# Then explain:
# Why can we concatenate along the feature dimension but not arbitrarily along every dimension?

visual = np.random.randn(32, 128)   # 32 samples x 128 visual features
sensor = np.random.randn(32, 64)    # 32 samples x 64 sensor features
 
combined = np.concatenate([visual, sensor], axis=1)
print("Combined shape:", combined.shape)
```

```python
# Task 16 — Practical Mini Challenge: Sensor Dataset
# Create a synthetic dataset representing:
# 500 machines
# 60 time steps
# 8 sensors

# Therefore:

# X.shape = (500, 60, 8)

# Perform the following:
# Extract machine 10. 
# Extract the first 20 time steps. 
# Extract sensor 5 from all machines. 
# Calculate the average value of every sensor. 
# Calculate the average value at every time step. 
# Find the machine with the highest average sensor reading. 
# Normalize each sensor. 
# Split the tensor into: 
# Training: 400 machines 
# Testing: 100 machines 

X16 = np.random.randn(500, 60, 8) * 10 + 50   # machines, time, sensors
print('X16.shape:', X16.shape)
 
machine_10       = X16[10]
print(machine_10.shape)
first_20_steps      = X16[:, :20, :]
print(first_20_steps.shape)
sensor_5_all     = X16[:, :, 5]
print(sensor_5_all.shape) 
avg_per_sensor   = X16.mean(axis=(0, 1))
print(avg_per_sensor) 
avg_per_timestep = X16.mean(axis=(0, 2))
print(avg_per_timestep)  
machine_avg      = X16.mean(axis=(1, 2))
highest_machine     = np.argmax(machine_avg)
print('Highest average machine:', highest_machine)
 
mu16, sigma16 = X16.mean(axis=(0,1), keepdims=True), X16.std(axis=(0,1), keepdims=True)
X16_norm = (X16 - mu16) / sigma16
X_train, X_test = X16[:400], X16[400:]

print(X_train.shape, X_test.shape)
```
