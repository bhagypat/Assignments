```python
# Q1.Write a program to check whether Employee ID 1050 exists.
print("1050" in employees) 

# Q2.Ask the user to enter an Employee ID and display whether it exists.
emp_id = input("Enter Employee ID: ")
if emp_id in employees:
    print(f"Employee ID {emp_id} exists.")
else:
    print(f"Employee ID {emp_id} does not exist.")
# # Q3.If the Employee ID exists, display all the employee details.

if emp_id in employees:
    print(employees[emp_id])

# # Q4. If the Employee ID does not exist, display: Employee not found.
if emp_id not in employees:
    print("Employee not found.")
    
# Q5.Display the salary of Employee 1032.
print(employees["1032"]["Salary"]) 

# Q6.Display the department of Employee 1087.
print(employees["1087"]["Department"]) 

# Q7.Print only the email address of Employee 1015.
print(employees["1015"]["Email"]) 

# Q8.Find all employees whose email address is missing (None).
not_recorded = [id for id, data in employees.items() if data["Email"] is None]
print(not_recorded)     

# Q9.Find all employees whose phone number is missing.
not_recorded = [id for id, data in employees.items() if data["Phone"] is None]
print(not_recorded)     

# Q10.Count how many employees have incomplete records (missing email or phone).
incomplete_count = sum(1 for details in employees.values() if details["Email"] is None or details["Phone"] is None)
print(incomplete_count)       

# Q11.Replace all missing email addresses with: Not Available
for emp_id, details in employees.items():
    if details["Email"] is None:
        details["Email"] = "Not Available"

print(employees["1007"]["Email"])

# Q12.Find all employees working in the IT department.
it_employees = [details["Name"] for details in employees.values() if details["Department"] == "IT"]
print(it_employees)

# Q13.Find all employees working in the Finance department.
finance_employees = [details["Name"] for details in employees.values() if details["Department"] == "Finance"]
print(finance_employees)

# Q14.Display the names of all employees managed by Manager_4.
finance_employees = [details["Name"] for details in employees.values() if details["Manager"] == "Manager_4"]
print(finance_employees)
# Q15.Find all employees older than 50 years.
finance_employees = [details["Name"] for details in employees.values() if details["Age"] > 50]
print(finance_employees)

# Q16.Find all employees younger than 30 years.
finance_employees = [details["Name"] for details in employees.values() if details["Age"] < 30]
print(finance_employees)

# Q17.Find the employee with the highest salary.
highest_paid = max(employees.values(), key=lambda x: x["Salary"])
print(highest_paid["Name"], highest_paid["Salary"]) 

# Q18.Find the employee with the lowest salary.
lowest_paid = min(employees.values(), key=lambda x: x["Salary"])
print(lowest_paid["Name"], lowest_paid["Salary"]) 

# Q19.Calculate the average salary of all employees.
average_salary = sum(details["Salary"] for details in employees.values()) / len(employees)
print(round(average_salary, 2))   

# Q20.Display employees earning more than ₹100,000.
above_100k = [details["Name"] for details in employees.values() if details["Salary"] > 100000]
print(above_100k) 
# Q21.Display employees earning less than ₹50,000.
below_50k = [details["Name"] for details in employees.values() if details["Salary"] < 50000]
print(below_50k)   
# Q22.Count how many employees belong to each department.

# Example output

# HR : 18
# Finance : 15
# IT : 22
# Sales : 17
# Marketing : 14
# Operations : 14

dept_count = {}
for details in employees.values():
    dept = details["Department"]
    dept_count[dept] = dept_count.get(dept, 0) + 1

for dept, count in dept_count.items():
    print(f"{dept} : {count}")
# Q23.Count how many employees are managed by each manager.
manager_count = {}
for details in employees.values():
    mgr = details["Manager"]
    manager_count[mgr] = manager_count.get(mgr, 0) + 1

for mgr, count in manager_count.items():
    print(f"{mgr} : {count}")
    
# Q24. Find which department has the largest number of employees.
dept_count = {}
for details in employees.values():
    dept = details["Department"]
    dept_count[dept] = dept_count.get(dept, 0) + 1

largest_dept = max(dept_count, key=dept_count.get)
print(largest_dept, dept_count[largest_dept])   
# Q25. Increase the salary of every employee by 10%.
for details in employees.values():
    details["Salary"] = round(details["Salary"] * 1.10, 2)
print(employees["1001"]["Salary"])

# Q26. Update the department of Employee 1040 to Research.
employees["1040"]["Department"] = "Research"
print(employees["1040"]["Department"])

# Q27. Add a new employee to the dictionary.
employees["1101"] = {
    "Name": "Employee_1101",
    "Department": "IT",
    "Age": 29,
    "Salary": 60000,
    "Email": "employee1101@company.com",
    "Phone": "9123456789",
    "Manager": "Manager_5"
}
print(employees["1101"])
# Q28. Delete Employee 1018 from the dictionary.
del employees["1018"]
print("1018" in employees)

# Q29. Update the phone number of Employee 1075.
employees["1075"]["Phone"] = "9000011122"
print(employees["1075"]["Phone"]) 
```

```python
# Generating a report  TRY THIS OUT, else we will see it in class

# Q30.Generate a report (HINT - USE -> 'print') showing:

#     Total Employees
#     Employees with Missing Email
#     Employees with Missing Phone
#     Average Salary
#     Highest Salary
#     Lowest Salary
total_employees = len(employees)
missing_email_count = sum(1 for d in employees.values() if d["Email"] is None)
missing_phone_count = sum(1 for d in employees.values() if d["Phone"] is None)
average_salary = sum(d["Salary"] for d in employees.values()) / total_employees
highest_salary = max(d["Salary"] for d in employees.values())
lowest_salary = min(d["Salary"] for d in employees.values())

print("Total Employees:", total_employees)
print("Employees with Missing Email:", missing_email_count)
print("Employees with Missing Phone:", missing_phone_count)
print("Average Salary:", round(average_salary, 2))
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
# Q31. Generate a report listing all employees department-wise.

# Example

# IT
# -----
# Employee_1001
# Employee_1005
# Employee_1022

# Finance
# ---------
# Employee_1008
# Employee_1012
# ...

dept_wise = {}
for emp_id, details in employees.items():
    dept = details["Department"]
    dept_wise.setdefault(dept, []).append(details["Name"])

for dept, names in dept_wise.items():
    print(dept)
    print("-" * len(dept))
    for name in names:
        print(name)
    print()
# Q32. Generate a report showing employees sorted by salary (highest to lowest).
sorted_employees = sorted(employees.items(), key=lambda x: x[1]["Salary"], reverse=True)
for emp_id, details in sorted_employees:
    print(details["Name"], ":", details["Salary"])
# Q33. Find all employees whose names contain the word "25".
contains_25 = [details["Name"] for details in employees.values() if "25" in details["Name"]]
print(contains_25) 
# Q34. Find all employees whose salary lies between ₹60,000 and ₹80,000.
salary_range = [details["Name"] for details in employees.values() if 60000 <= details["Salary"] <= 80000]
print(salary_range) 
# Q35. Find managers supervising more than 8 employees.
manager_count = {}
for details in employees.values():
    mgr = details["Manager"]
    manager_count[mgr] = manager_count.get(mgr, 0) + 1

busy_managers = {mgr: count for mgr, count in manager_count.items() if count > 8}
print(busy_managers)
# Q36. Find the department with the highest average salary.
dept_salaries = {}
for details in employees.values():
    dept = details["Department"]
    dept_salaries.setdefault(dept, []).append(details["Salary"])

dept_avg_salary = {dept: sum(salaries)/len(salaries) for dept, salaries in dept_salaries.items()}
highest_avg_dept = max(dept_avg_salary, key=dept_avg_salary.get)
print(highest_avg_dept, round(dept_avg_salary[highest_avg_dept], 2))
# Q37. Identify employees with both email and phone number missing.
both_missing = [details["Name"] for details in employees.values() if details["Email"] is None and details["Phone"] is None]
print(both_missing)
# Q38. Find duplicate phone numbers (if any).
phone_count = {}
for details in employees.values():
    phone = details["Phone"]
    if phone is not None:
        phone_count[phone] = phone_count.get(phone, 0) + 1

duplicates = {phone: count for phone, count in phone_count.items() if count > 1}
print(duplicates)
# Q39. Find employees whose email domain is company.com.
company_domain = [details["Name"] for details in employees.values()
                   if details["Email"] is not None and details["Email"].endswith("@company.com")]
print(company_domain)  
# Q40. Create a new dictionary containing only employees from the IT department.
it_only = {emp_id: details for emp_id, details in employees.items() if details["Department"] == "IT"}
print(it_only)  
```
